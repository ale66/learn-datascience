##### Ingest a network from an SVG file with the optimised node positioning

```python
from bs4 import BeautifulSoup
```

To run this notebook in Google Colab, we first need to upload the SVG file.

```python
DATFILE = "./data/test_graph.dat"
```

```python
def graph_from_svg(stream) -> tuple:
    """Create a Networkx network by reading from an SVG file."""
    graph = nx.Graph()

    attrs = {
        "line" :  ["x1", "y1", "x2", "y2"]
    }

    op = open(stream, "r")
    xml = op.read()
    soup = BeautifulSoup(xml, features="xml")

    counter = 0
    node_dict = {}
    pos_dict = {}

    for attr in attrs.keys():
        tmps = soup.find_all(attr)

        for t in tmps:
            node1 = (t['x1'], t['y1'])
            node2 = (t['x2'], t['y2'])

            if not node1 in node_dict:
                node_dict[node1] = str(counter)
                pos_dict[str(counter)] = (float(node1[0]), float(node1[1]))
                counter += 1

            if not node2 in node_dict:
                node_dict[node2] = str(counter)
                pos_dict[str(counter)] = (float(node2[0]), float(node2[1]))
                counter += 1

            graph.add_edge(node_dict[node1], node_dict[node2])

    # Save the graph in an edge list format
    nx.write_edgelist(graph, DATFILE, data=False)

    return graph, pos_dict
```

#### Define a function that calculates the distance from a root node

```python
def node_distance(graph, root_node) -> list:
    queue = [root_node]
    list_distances = []

    # reset the old keys
    if 'distance' in graph.nodes[root_node]:
        for node in graph.nodes():
            del graph.nodes[node]['distance']

    graph.nodes[root_node]["distance"] = 0

    # now...
    while len(queue):
        working_node = queue.pop(0)

        for node in graph.neighbors(working_node):

            if len(graph.nodes[node]) == 0:
                graph.nodes[node]["distance"] = \
                    graph.nodes[working_node]["distance"] + 1

                queue.append(node)

    for node in graph.nodes():
        list_distances.append(((root_node, node), graph.nodes[node]["distance"]))

    return list_distances
```

#### Closeness Centrality

```python
norm = 0.0
diz_c = {}
l_values = []

for n in G.nodes():
    l = node_distance(G, n)

    ave_length = 0

    for path in l:
        ave_length += float(path[1]) / (G.number_of_nodes() - 1 - 0)

    norm += 1 / ave_length

    diz_c[n] = 1 / ave_length

    l_values.append(diz_c[n])

# visualization
nx.draw_networkx_edges(G, pos)
for n in G.nodes():

    list_nodes = [n]

    pink_to_red = mcolors.LinearSegmentedColormap.from_list("pink_red", ["#ffc0cb", "#8b0000"])

    color_values = []
    norm_val = (diz_c[n] - min(l_values)) / (max(l_values) - min(l_values))
    color_values.append(norm_val)

    nx.draw_networkx_nodes(
    G,
    {n:pos[n]}, [n],
    node_size=100,
    node_color=color_values,
    cmap=pink_to_red,
    vmin=0, vmax=1)

#plt.savefig("./data/closeness_200.png",dpi=200)
```

#### Betweenness Centrality

```python
list_of_nodes = list(G.nodes())

num_of_nodes = G.number_of_nodes()

bc = {}  # we will need this dictionary later on

for i in range(0, num_of_nodes-1):
    for j in range(i+1, num_of_nodes):
        # print(list_of_nodes[i])
        # print(list_of_nodes[j])

        paths = nx.all_shortest_paths(
            G=G,
            source=list_of_nodes[i],
            target=list_of_nodes[j]
        )

        count = 0.0

        path_dict = {}

        for p in paths:
            # print p
            count += 1.0
            for n in p[1:-1]:
                if not n in path_dict:
                    path_dict[n] = 0.0
                path_dict[n] += 1.0

        for n in path_dict.keys():
            path_dict[n] = path_dict[n] / count

            if not n in bc:
                bc[n] = 0.0

            bc[n] += path_dict[n]
```

#### Giant Component through a Breadth-First Search

```python
def giant_component_size(input_graph) -> tuple:

    graph = input_graph.copy()

    components = []
    # print(graph)

    node_list = list(graph.nodes())
    # print(node_list.type)

    while len(node_list) != 0:
        # print(list(node_list))
        # print(node_list[0])
        root_node = node_list[0]
        component_list = [root_node]
        queue = [root_node]

        graph.nodes[root_node]["visited"] = True

        while len(queue):

            working_node = queue.pop(0)

            for node in graph.neighbors(working_node):
                # check if any node attribute exists
                if len(graph.nodes[node]) == 0:
                    graph.nodes[node]["visited"] = True
                    queue.append(node)
                    component_list.append(node)

        components.append((len(component_list), component_list))

        # remove the nodes of the component just discovered
        for node in component_list: node_list.remove(node)

    components.sort(reverse = True)

    giant_component = components[0][1]

    # this might become useful later
    # size_giant_component = components[0][0]

    return giant_component, len(components)
```

```python

GCC, num_components = giant_component_size(G_test)

print ("Giant Connected Component:", GCC)

print ("Number of components:", num_components)
```
